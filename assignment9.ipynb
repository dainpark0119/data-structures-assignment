```python
import matplotlib.pyplot as plt
import networkx as nx
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

locations = ["빠오즈푸", "이쿠", "마복림할머니집", "트라토리아 진", "다온카츠", "익선취향", "땀땀", "담금"]
adj_matrix = [
    [0, 3.5, 1.1, 3.5, 4.5, 2, 2, 2.3],
    [3.5, 0, 3.6, 7, 5.5, 4, 2, 4.7],
    [1.1, 3.6, 0, 3, 3.5, 0.8, 2.3, 3.5],
    [3.5, 7, 3, 0, 4, 3, 5, 5],
    [4.5, 5.5, 3.5, 4, 0, 2.7, 5, 7],
    [2, 4, 0.8, 3, 2.7, 0, 3, 4],
    [2, 2, 2.3, 5, 5, 3, 0, 3],
    [2.3, 4.7, 3.5, 5, 7, 4, 3, 0]
]

G = nx.Graph()
for i in range(len(locations)):
    for j in range(i + 1, len(locations)):
        G.add_edge(locations[i], locations[j], weight=adj_matrix[i][j])

pos = {
    "빠오즈푸": (0.6, 0.4), "이쿠": (0.2, 0.1), "마복림할머니집": (0.45, 0.45),
    "트라토리아 진": (0.5, 0.9), "다온카츠": (0.1, 0.7), "익선취향": (0.35, 0.55),
    "땀땀": (0.55, 0.2), "담금": (0.9, 0.3)
}

dfs_visited = []
dfs_edges = []

def run_dfs(u, p=None):
    if u not in dfs_visited:
        dfs_visited.append(u)
        if p:
            dfs_edges.append((p, u))
        neighbors = sorted(G.neighbors(u), key=lambda x: G[u][x]['weight'])
        for v in neighbors:
            if v not in dfs_visited:
                run_dfs(v, u)

run_dfs("빠오즈푸")

fig, ax = plt.subplots(figsize=(12, 9))

def update(frame):
    ax.clear()
    ax.set_title(f"DFS Traversal: {dfs_visited[frame]}", fontsize=16)
    
    nx.draw_networkx_nodes(G, pos, node_color='whitesmoke', edgecolors='gray', node_size=2000, ax=ax)
    nx.draw_networkx_labels(G, pos, font_size=10, font_family='Malgun Gothic', ax=ax)
    nx.draw_networkx_edges(G, pos, edge_color='lightgray', alpha=0.3, ax=ax)
    
    edge_labels = nx.get_edge_attributes(G, 'weight')
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=9, ax=ax)
    
    current_nodes = dfs_visited[:frame+1]
    current_edges = dfs_edges[:frame]
    
    nx.draw_networkx_nodes(G, pos, nodelist=current_nodes, node_color='orange', node_size=2100, ax=ax)
    if frame > 0:
        nx.draw_networkx_edges(G, pos, edgelist=current_edges, edge_color='red', width=3, ax=ax)

ani = FuncAnimation(fig, update, frames=len(dfs_visited), interval=1000, repeat=False)
plt.close()

HTML(ani.to_jshtml())
```

```python
import matplotlib.pyplot as plt
import networkx as nx
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
from collections import deque

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

locations = ["빠오즈푸", "이쿠", "마복림할머니집", "트라토리아 진", "다온카츠", "익선취향", "땀땀", "담금"]
adj_matrix = [
    [0, 3.5, 1.1, 3.5, 4.5, 2, 2, 2.3],
    [3.5, 0, 3.6, 7, 5.5, 4, 2, 4.7],
    [1.1, 3.6, 0, 3, 3.5, 0.8, 2.3, 3.5],
    [3.5, 7, 3, 0, 4, 3, 5, 5],
    [4.5, 5.5, 3.5, 4, 0, 2.7, 5, 7],
    [2, 4, 0.8, 3, 2.7, 0, 3, 4],
    [2, 2, 2.3, 5, 5, 3, 0, 3],
    [2.3, 4.7, 3.5, 5, 7, 4, 3, 0]
]

G = nx.Graph()
for i in range(len(locations)):
    for j in range(i + 1, len(locations)):
        G.add_edge(locations[i], locations[j], weight=adj_matrix[i][j])

pos = {
    "빠오즈푸": (0.6, 0.4), "이쿠": (0.2, 0.1), "마복림할머니집": (0.45, 0.45),
    "트라토리아 진": (0.5, 0.9), "다온카츠": (0.1, 0.7), "익선취향": (0.35, 0.55),
    "땀땀": (0.55, 0.2), "담금": (0.9, 0.3)
}

bfs_visited = ["빠오즈푸"]
bfs_edges = []
queue = deque(["빠오즈푸"])

while queue:
    u = queue.popleft()
    neighbors = sorted(G.neighbors(u), key=lambda x: G[u][x]['weight'])
    for v in neighbors:
        if v not in bfs_visited:
            bfs_visited.append(v)
            bfs_edges.append((u, v))
            queue.append(v)

fig, ax = plt.subplots(figsize=(12, 9))

def update(frame):
    ax.clear()
    ax.set_title(f"BFS Traversal: {bfs_visited[frame]}", fontsize=16)
    
    nx.draw_networkx_nodes(G, pos, node_color='whitesmoke', edgecolors='gray', node_size=2000, ax=ax)
    nx.draw_networkx_labels(G, pos, font_size=10, font_family='Malgun Gothic', ax=ax)
    nx.draw_networkx_edges(G, pos, edge_color='lightgray', alpha=0.3, ax=ax)
    
    edge_labels = nx.get_edge_attributes(G, 'weight')
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=9, ax=ax)
    
    current_nodes = bfs_visited[:frame+1]
    current_edges = bfs_edges[:frame]
    
    nx.draw_networkx_nodes(G, pos, nodelist=current_nodes, node_color='skyblue', node_size=2100, ax=ax)
    if frame > 0:
        nx.draw_networkx_edges(G, pos, edgelist=current_edges, edge_color='blue', width=3, ax=ax)

ani = FuncAnimation(fig, update, frames=len(bfs_visited), interval=1000, repeat=False)
plt.close()

HTML(ani.to_jshtml())
```

```python
import matplotlib.pyplot as plt
import networkx as nx
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import heapq

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

locations = ["빠오즈푸", "이쿠", "마복림할머니집", "트라토리아 진", "다온카츠", "익선취향", "땀땀", "담금"]
adj_matrix = [
    [0, 3.5, 1.1, 3.5, 4.5, 2, 2, 2.3],
    [3.5, 0, 3.6, 7, 5.5, 4, 2, 4.7],
    [1.1, 3.6, 0, 3, 3.5, 0.8, 2.3, 3.5],
    [3.5, 7, 3, 0, 4, 3, 5, 5],
    [4.5, 5.5, 3.5, 4, 0, 2.7, 5, 7],
    [2, 4, 0.8, 3, 2.7, 0, 3, 4],
    [2, 2, 2.3, 5, 5, 3, 0, 3],
    [2.3, 4.7, 3.5, 5, 7, 4, 3, 0]
]

G = nx.Graph()
for i in range(len(locations)):
    for j in range(i + 1, len(locations)):
        G.add_edge(locations[i], locations[j], weight=adj_matrix[i][j])

pos = {
    "빠오즈푸": (0.6, 0.4), "이쿠": (0.2, 0.1), "마복림할머니집": (0.45, 0.45),
    "트라토리아 진": (0.5, 0.9), "다온카츠": (0.1, 0.7), "익선취향": (0.35, 0.55),
    "땀땀": (0.55, 0.2), "담금": (0.9, 0.3)
}

prim_visited = ["빠오즈푸"]
prim_edges = []
edges_heap = []

for neighbor in G.neighbors("빠오즈푸"):
    heapq.heappush(edges_heap, (G["빠오즈푸"][neighbor]['weight'], "빠오즈푸", neighbor))

while edges_heap and len(prim_visited) < len(locations):
    weight, u, v = heapq.heappop(edges_heap)
    if v not in prim_visited:
        prim_visited.append(v)
        prim_edges.append((u, v))
        for neighbor in G.neighbors(v):
            if neighbor not in prim_visited:
                heapq.heappush(edges_heap, (G[v][neighbor]['weight'], v, neighbor))

fig, ax = plt.subplots(figsize=(12, 9))

def update(frame):
    ax.clear()
    ax.set_title(f"Prim's Algorithm (MST): {prim_visited[frame]} 연결 완료", fontsize=16)
    
    nx.draw_networkx_nodes(G, pos, node_color='whitesmoke', edgecolors='gray', node_size=2000, ax=ax)
    nx.draw_networkx_labels(G, pos, font_size=10, font_family='Malgun Gothic', ax=ax)
    nx.draw_networkx_edges(G, pos, edge_color='lightgray', alpha=0.3, ax=ax)
    
    edge_labels = nx.get_edge_attributes(G, 'weight')
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=9, ax=ax)
    
    current_nodes = prim_visited[:frame+1]
    current_edges = prim_edges[:frame]
    
    nx.draw_networkx_nodes(G, pos, nodelist=current_nodes, node_color='lightgreen', node_size=2100, ax=ax)
    if frame > 0:
        nx.draw_networkx_edges(G, pos, edgelist=current_edges, edge_color='green', width=4, ax=ax)

ani = FuncAnimation(fig, update, frames=len(prim_visited), interval=1000, repeat=False)
plt.close()

HTML(ani.to_jshtml())
```

```python
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

G = nx.Graph()

edges = [
    ("빠오즈푸","이쿠",3.5),
    ("빠오즈푸","마복림",1.1),
    ("빠오즈푸","트라토리아진",3.5),
    ("빠오즈푸","다운카츠",4.5),
    ("빠오즈푸","익선취향",2),
    ("빠오즈푸","땀땀",2),
    ("빠오즈푸","담금",2.3),

    ("이쿠","마복림",3.6),
    ("이쿠","트라토리아진",7),
    ("이쿠","다운카츠",5.5),
    ("이쿠","익선취향",4),
    ("이쿠","땀땀",2),
    ("이쿠","담금",4.7),

    ("마복림","트라토리아진",3),
    ("마복림","다운카츠",3.5),
    ("마복림","익선취향",0.8),
    ("마복림","땀땀",2.3),
    ("마복림","담금",3.5),

    ("트라토리아진","다운카츠",4),
    ("트라토리아진","익선취향",3),
    ("트라토리아진","땀땀",5),
    ("트라토리아진","담금",5),

    ("다운카츠","익선취향",2.7),
    ("다운카츠","땀땀",5),
    ("다운카츠","담금",7),

    ("익선취향","땀땀",3),
    ("익선취향","담금",4),

    ("땀땀","담금",3)
]

G.add_weighted_edges_from(edges)

pos = {
    "다운카츠":(-3,3),
    "트라토리아진":(0,4),
    "익선취향":(-1,1),
    "마복림":(0,0),
    "빠오즈푸":(2,-1),
    "땀땀":(2,-4),
    "담금":(5,-1),
    "이쿠":(-2,-5)
}

edges_sorted = sorted(G.edges(data=True), key=lambda x:x[2]['weight'])

parent = {n:n for n in G.nodes()}

def find(x):
    if parent[x] == x:
        return x
    return find(parent[x])

def union(a,b):
    parent[find(a)] = find(b)

mst = []
frames = [[]]

for u,v,d in edges_sorted:
    if find(u) != find(v):
        union(u,v)
        mst.append((u,v))
        frames.append(list(mst))

fig, ax = plt.subplots(figsize=(12,9))

def update(frame):
    ax.clear()

    nx.draw(
        G,
        pos,
        with_labels=True,
        node_color='whitesmoke',
        edge_color='lightgray',
        node_size=2500,
        font_family='Malgun Gothic',
        ax=ax
    )

    nx.draw_networkx_edge_labels(
        G,
        pos,
        edge_labels=nx.get_edge_attributes(G,'weight'),
        font_color='blue',
        ax=ax
    )

    nx.draw_networkx_edges(
        G,
        pos,
        edgelist=frames[frame],
        edge_color='orange',
        width=4,
        ax=ax
    )

    ax.set_title(f"Kruskal Step {frame}")

ani = FuncAnimation(fig, update, frames=len(frames), interval=1200, repeat=False)

plt.close()

HTML(ani.to_jshtml())
```
