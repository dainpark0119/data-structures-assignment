1. 해싱 충돌 해결 방법은 개방 주소 방법과 체이닝이 있다. 
개방 주소 방법: 장점은 추가적인 메모리 공간이 필요없다는 것이고 단점은 데이터가 많아질수록 충돌이 잦아져 성능이 저하된다는 것이다. 
체이닝: 장점은 해시 테이블의 크기보다 더 많은 데이터를 저장할 수 있고 삭제 작업이 상대적으로 간단하다는 것이고 단점은 연결 리스트를 위한 추가 메모리가 필요하다는 것이다. 

2-1. 테이블 크기가 1,024일 때, divisor로 1,024($2^{10}$)를 사용하면 키 $x$의 하위 10비트에 의해서만 주소가 결정된다.

만약 데이터의 하위 비트들이 특정 패턴을 가지고 있다면, 주소가 고르게 분산되지 않고 심각한 충돌이 발생 할 수 있다. 일반적으로는 소수를 테이블 크기로 사용하는 것이 충돌을 줄이는 데 유리하다.

2-2. 해시 함수의 가장 중요한 특징은 결정론적이어야 한다. 즉, 동일한 입력 값 $x$에 대해서는 항상 동일한 해시 값이 나와야 한다. 위 함수는 random() 함수를 포함하고 있어 동일한 $x$를 입력해도 호출할 때마다 다른 주소를 반환할 수 있다. 이 경우 데이터를 저장한 위치를 다시 찾을 수 없어 해시 함수로 사용할 수 없다.

2-7. 삽입 효율이 떨어진다. 리스트의 맨 앞에 삽입하면 O(1)의 시간이 걸리지만, 맨 뒤에 삽입하려면 리스트 전체를 탐색해야 해서 O(n)의 시간이 소요된다.

2-8. 
삽입: 정렬된 위치를 찾아야해서 O(n)으로 효율성이 떨어지거나 비슷하다. 

삭제: 삭제할 원소를 찾는 과정이 필요해서 정렬 여부와 상관없이 O(n)이다.

성공적인 검색: 찾고자 하는 원소를 만날 때까지 탐색해서 정렬되지 않은 경우와 큰 차이가 없다.

실패하는 검색: 효율성이 높아진다. 정렬되어 있으면 찾으려는 키보다 큰 값을 만나는 순간 검색을 중단할 수 있어 리스트 전체를 끝까지 보지 않아도 되기 때문이다.
